## Cleaning the data


```python
import pandas as pd 
```

```python
df = pd.read_csv("data/sales_data.csv",quoting=3)
```

```python
df = df.dropna(subset=["quantity"])
```

```python
df = df[df["quantity"] >= 0]
```

```python
df["unit_price"] = pd.to_numeric(df["unit_price"], errors="coerce")
```

```python
df = df.dropna(subset=["unit_price"])
```

```python
df = df.drop_duplicates()
```

```python
df.columns = df.columns.str.strip('"')
```

```python
df.iloc[:, 0] = df.iloc[:, 0].str.strip('"')
```

```python
df["date"]  = pd.to_datetime(df["date"], format="mixed")
```

```python
df["category"] = df["category"].str.strip().str.title()
```

```python
df["store"] = df["store"].str.strip().str.title()
```

```python
print("DATA  QUALITY REPORT")
print("Starting rows:", 600)
print("Row removed - missing quantity:", 1)
print("Row removed - invalid price:", 1)
print("Row removed - negative quantity:", 1)
print("Row removed - duplicate:", 1)
print("Total rows removed:", 4)
print("Clean rows remaining:", len(df))
```

## Descriptive Analysis

```python
df["revenue"] = df["quantity"] * df["unit_price"]
```

```python
print(f" The total revenue generated across all stores: {df['revenue'].sum()}")

## The total revenue generated across all stores during the period was R324303.This represents the combined value of all sales transactions after cleaning the data
```

```python
df.groupby("category")["revenue"].sum().sort_values(ascending=False)

## Beverages generated the highest revenue at R109320 followed by Meat at R104055.Bakery generated the lowest revenue at R22518.This shows that revenue was concentrated more heavily in the Beverages and Meat categories than in Produce and Bakery.
```

```python
df.groupby("store")["revenue"].sum().sort_values(ascending=False)

## Sandton generated the highest revenue at R89137.50, while Rosebank generated the lowest at R67184.50. This shows that the stores did not generate the same amount of revenue during the period
```

```python
df.groupby("product_name")["quantity"].sum().sort_values(ascending=False)

## Rooibos Tea 100g was the best-selling producct by quantity, with 570 units sold.This means it had the highest sales volume among the products in the dataset.
```

```python
df.groupby("product_name")["revenue"].sum().sort_values(ascending=False)

## Coffee Beans 1kg generated the highest revenue at R77760, This means it contributed more revenue than any other individual product in the dataset.
```

```python
## Rooibos Tea = highest quantity
## Coffee Beans = highest revenue
## The product sold in the largest quantity was not the product generating the most revenue.This suggests that sales volume and revenue are influenced by both the number of units sold and the price per unit.
```

```python
df["month"] = df["date"].dt.to_period("M")
```

```python
df.groupby("month")["revenue"].sum()
```

```python
## Revenue flactuated throughout the period.It increased from January to February, decreased in March, then increased substantially through April and reached its highest point in May at R77000. Revenue decreased in June and dropped shaply in July to R2728.
```

```python
df["revenue"].mean()

## The average transaction value was approximately R544.13. This means that, across the cleaned dataset, the average transaction generated about R544.13 in revenue.
```

```python
df["payment_method"].value_counts()

## Mobile was the most commonly used payment method, with 166 transactions. This means more transactions were paid using Mobile than EFT, Cash, or Card in this dataset.
```
